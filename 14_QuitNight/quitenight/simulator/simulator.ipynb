```python
import time
import json
import random
import paho.mqtt.client as mqtt

# === MQTT Configuration ===
MQTT_BROKER = "b6e0e48a748c4d20971053ddf587142e.s1.eu.hivemq.cloud"
MQTT_PORT = 8883
MQTT_USERNAME = "espProject"
MQTT_PASSWORD = "Esp12345678"
MQTT_TOPIC = "quietnight/esp32/data"

# === MQTT Client Setup ===
client = mqtt.Client()
client.username_pw_set(MQTT_USERNAME, MQTT_PASSWORD)
client.tls_set()  # Enable SSL for port 8883

client.connect(MQTT_BROKER, MQTT_PORT)
client.loop_start()

print(f"📡 Publishing to topic: {MQTT_TOPIC}")

try:
    while True:
        # Generate synthetic sensor values
        temperature_c = round(random.uniform(28.0, 32.0), 2)
        temperature_f = round((temperature_c * 9 / 5) + 32, 2)
        humidity = round(random.uniform(60.0, 90.0), 2)
        heat_index_c = round(temperature_c + random.uniform(3, 7), 2)
        heat_index_f = round((heat_index_c * 9 / 5) + 32, 2)
        vibration_analog = random.randint(0, 1023)
        vibration_digital = random.choice([0, 1])
        light_lux = round(random.uniform(0, 50), 2)

        payload = {
            "temperature_c": temperature_c,
            "temperature_f": temperature_f,
            "humidity": humidity,
            "heat_index_c": heat_index_c,
            "heat_index_f": heat_index_f,
            "vibration_analog": vibration_analog,
            "vibration_digital": vibration_digital,
            "light_lux": light_lux,
            "source": "simulator"  # ✅ Identifies this as simulated data
        }

        json_data = json.dumps(payload)
        client.publish(MQTT_TOPIC, json_data)
        print("✅ Sent:", json_data)
        time.sleep(5)  # Send every 5 seconds

except KeyboardInterrupt:
    print("🛑 Simulator stopped.")
    client.loop_stop()
    client.disconnect()
```
